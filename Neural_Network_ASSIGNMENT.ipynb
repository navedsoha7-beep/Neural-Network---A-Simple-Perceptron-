2. Explain the basic architecture and functioning of a Perceptron. What are its
limitations?




     - A Perceptron is the simplest type of artificial neural network unit. It mimics how a biological neuron works.

The main limitation of a perceptron is that it can only solve linearly separable problems. It fails on tasks like XOR, cannot model complex decision boundaries, and struggles with multi-class or non-linear data



3.  Describe the purpose of activation function in neural networks. Compare
Sigmoid, ReLU, and Tanh functions.




     - - Sigmoid: Good for output layers in binary classification, but poor for deep networks due to vanishing gradients.
- Tanh: Improves over sigmoid by being zero-centered, but still struggles with vanishing gradients in deep networks.
- ReLU: The most widely used activation in modern deep learning because it is simple, efficient, and mitigates vanishing gradients. However, it can cause "dead neurons" if weights update poorly.






4.  What is the difference between Loss function and Cost function in neural
networks? Provide examples.




     - Loss function measures error for the single data while cost function measures error for the entire data.




- During training, the model computes loss per sample.
- Then it aggregates into a cost function (average loss) to update weights using optimization algorithms like gradient descent.






5.  What is the role of optimizers in neural networks? Compare Gradient
Descent, Adam, and RMSprop.




    - Optimizers in neural networks adjust the model’s weights to minimize the cost (loss) function. They determine how the network learns during training. Gradient Descent is the simplest optimizer, while Adam and RMSprop are advanced variants that adapt learning rates for faster and more stable convergence

```python
#7. Implement and visualize Sigmoid, ReLU, and Tanh activation functions using Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)

# Generate input values
x = np.linspace(-10, 10, 400)

# Compute outputs
y_sigmoid = sigmoid(x)

y_relu = relu(x)
y_tanh = tanh(x)

# Plot all functions
plt.figure(figsize=(10, 6))

plt.plot(x, y_sigmoid, label="Sigmoid", color="blue")
plt.plot(x, y_relu, label="ReLU", color="green")
plt.plot(x, y_tanh, label="Tanh", color="red")

plt.title("Activation Functions: Sigmoid, ReLU, Tanh")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```
